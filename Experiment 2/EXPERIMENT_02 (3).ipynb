```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("liberaries are installed")
```

```python
df = pd.read_csv("/workspaces/Mastery_in_DataAnalytics_and_Visualizations_and_Career_Advancemen/WA_Fn-UseC_-HR-Employee-Attrition-selected-columns.csv")

print(df.head())
```

```python
print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nSummary Statistics:")
print(df.describe(include="all"))
```

```python
for column in df.columns:
    print("\n", column)
    print(df[column].value_counts().head(10))
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numerical_columns)
```

```python
df[numerical_columns].hist(
    figsize=(12, 10),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 5))
    sns.kdeplot(data=df, x=column, fill=True)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Density")
    plt.show()
```

```python
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical Columns:")
print(categorical_columns)
```

```python
for column in categorical_columns:
    plt.figure(figsize=(10, 5))
    
    sns.countplot(
        data=df,
        x=column,
        order=df[column].value_counts().head(10).index
    )
    
    plt.title(f"Top Categories in {column}")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
correlation = df[numerical_columns].corr()

print(correlation)
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 5))
    
    sns.boxplot(data=df, y=column)
    
    plt.title(f"Box Plot of {column}")
    plt.show()
```

```python
if len(numerical_columns) >= 2:
    
    x = numerical_columns[0]
    y = numerical_columns[1]
    
    plt.figure(figsize=(8, 5))
    
    sns.scatterplot(
        data=df,
        x=x,
        y=y
    )
    
    plt.title(f"{x} vs {y}")
    plt.xlabel(x)
    plt.ylabel(y)
    plt.show()
```

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 5))
    sns.histplot(data=df, x=column, kde=True)
    plt.title(f"Univariate Analysis - {column}")
    plt.show()
```

```python
if len(numerical_columns) >= 2:
    
    x = numerical_columns[0]
    y = numerical_columns[1]
    
    plt.figure(figsize=(8, 5))
    
    sns.scatterplot(
        data=df,
        x=x,
        y=y
    )
    
    plt.title(f"Bivariate Analysis - {x} vs {y}")
    plt.show()
```

```python
if len(numerical_columns) >= 3:
    
    sns.pairplot(
        df[numerical_columns].dropna()
    )
    
    plt.show()
```

```python
correlation_matrix = df[numerical_columns].corr()

print("Correlation with numerical variables:")
print(correlation_matrix)
```

```python
for column in correlation_matrix.columns:
    for row in correlation_matrix.index:
        
        if column != row:
            value = correlation_matrix.loc[row, column]
            
            if abs(value) >= 0.7:
                print(
                    f"{row} and {column}: "
                    f"Correlation = {value:.2f}"
                )
```
